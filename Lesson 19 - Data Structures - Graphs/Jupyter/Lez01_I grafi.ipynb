# RICERCA POZZO UNIVERSALE

```python
MM=[[]]

MM.append([0,1,1,0,0]) # Riga di adiacenza del nodo 0
MM.append([0,0,1,0,0]) # Riga di adiacenza del nodo 1
MM.append([0,0,0,0,0]) # Riga di adiacenza del nodo 2
MM.append([0,0,1,0,0]) # Riga di adiacenza del nodo 3
MM.append([0,1,1,0,0]) # Riga di adiacenza del nodo 4
```

```python
# ALGORITMO di RICERCA in O(n^2)
def pozzoUniversale1(M):
    for i in range(0,len(M)):
        if (pozzo(i,M)): return i
    return "N/A"

pozzoUniversale1(MM)


# ALGORITMO di RICERCA in Θ(n)
def pozzoUniversale2(M):
    p=0
    for i in range(0,len(M)):
        if (M[p][i]==1):
            p=i
    x=pozzo(p,M)
    if x: return p
    return "N\A"

pozzoUniversale2(MM)
```
